```python
import numpy as np

# 1. Define the 2x2 matrix A
A = np.array([[4, 11],
              [14, 5]], dtype=float)

# 2. Compute the SVD
# np.linalg.svd returns U, a 1D array of singular values, and V^T
U, s, Vt = np.linalg.svd(A)

# 3. Convert the 1D singular values array into a 2x2 diagonal matrix (Sigma)
Sigma = np.diag(s)

# 4. Display the results nicely
print("--- Matrix U (Left Singular Vectors) ---")
print(np.round(U, 4))

print("\n--- Matrix Sigma (Singular Values Diagonal) ---")
print(np.round(Sigma, 4))

print("\n--- Matrix Vt (Right Singular Vectors Transposed) ---")
print(np.round(Vt, 4))

# 5. Optional Verification: Multiply U * Sigma * Vt to reconstruct A
A_reconstructed = U @ Sigma @ Vt
print("\n--- Verification (U * Sigma * Vt) ---")
print(np.round(A_reconstructed, 4))
```

```python
import numpy as np

# 1. Define a 3x3 matrix A
A = np.array([
    [3.0, 2.0, 2.0],
    [2.0, 3.0, -2.0],
    [1.0, 1.0, 2.0]
], dtype=float)

# 2. Compute the SVD
# np.linalg.svd returns U (3x3), s (1x3 array of singular values), and Vt (3x3)
U, s, Vt = np.linalg.svd(A)

# 3. Convert the 1D singular values array into a 3x3 diagonal matrix (Sigma)
Sigma = np.diag(s)

# 4. Display the results rounded to 4 decimal places
print("--- Matrix U (3x3 Left Singular Vectors) ---")
print(np.round(U, 4))

print("\n--- Matrix Sigma (3x3 Singular Values Diagonal) ---")
print(np.round(Sigma, 4))

print("\n--- Matrix Vt (3x3 Right Singular Vectors Transposed) ---")
print(np.round(Vt, 4))

# 5. Verification: Reconstruct original matrix A
A_reconstructed = U @ Sigma @ Vt
print("\n--- Verification (U @ Sigma @ Vt) ---")
print(np.round(A_reconstructed, 4))
```
