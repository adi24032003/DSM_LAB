```python
import numpy as np
arr1= np.array([[6,3],[2,6]])
arr2= np.array([[6,3],[1,2]])
print(str(arr1+arr2)+"addition")
print("\n")
print(str(arr1-arr2)+"subsraction")
print("\n")
print(str(arr1*arr2)+"multiplication")
print("\n")
print(np.dot(arr1,arr2))
print("\n")
print(np.transpose(arr1))
```
