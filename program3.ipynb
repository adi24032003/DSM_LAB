```python
import matplotlib.pyplot as plt

x = [1,2,6,18]
y = [3,10,12,20]

plt.plot(x, y,linestyle=':',color ='red',marker='o')

plt.title("Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)


plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [2, 4, 6, 8, 10]


plt.subplot(1, 2, 1)
plt.plot(x, y1, marker='o', color='blue')
plt.title("Line Graph")


plt.subplot(1, 2, 2)
plt.bar(x, y2, color='green')
plt.title("Bar Graph")


plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
men=[22,30,35,35,26]
women=[25,32,30,35,29]
import matplotlib.pyplot as plt
x=[1,2,3,4,5]

plt.bar(x,men,width=0.4,label="men")
plt.bar([i+0.4 for i in x],women,width=0.4,label="women")

plt.xlabel("Group")
plt.ylabel("Scores")

plt.legend()


plt.show()

```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

plt.plot(x, y1, color='blue',  label='Line 1')

plt.plot(x, y2, color='red', label='Line 2')


plt.title("Multiple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

language=["Java","Python","PHP","JavaScript","C#","C++"]
popularity=[22.2,17.6,8.8,8,7.7,6.7]

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.pie(popularity,labels=language,autopct="%1.1f%%")

plt.subplot(1,3,2)
plt.scatter(language,popularity)
plt.xticks(rotation=45)

plt.subplot(1,3,3)
plt.barh(language,popularity)
plt.grid()

plt.tight_layout()
plt.show()
```
